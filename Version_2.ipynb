Function Planning Guide

Before coding, complete this planning table.

|Operation|Function Name|Parameters Needed|Process Inside the Function| Return Value|
|-----|-----|-----|-----|-----|
|Addition|add_numbers|num1, num2|add the two numbers|Sum|
|Subtraction|subtract_numbers|num1, num2|subtract the two numbers|Difference|
|Multiplication|multiply_numbers|num1, num2|multiply the two numbers|Product|
|Division|divide_numbers|num1, num2|divide the two numbers|Quotient|

```python
# Returns (sum, difference, product, quotient)
def calculate(num1, num2):
    total = num1 + num2
    difference = num1 - num2
    product = num1 * num2
    # Guard against division by zero
    quotient = num1 / num2 if num2 != 0 else "Cannot divide by zero."
    return total, difference, product, quotient


# Inputs (float() allows decimals)
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Unpack the four returned values
total, difference, product, quotient = calculate(num1, num2)

# Output
print("Sum:", total)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```
